```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

print("RecursiveCharacterTextSplitter OK")
```

```python
from pathlib import Path

markdown_file = Path("../data/processed/chap1.md")

print(markdown_file.exists())
```

```python
text = markdown_file.read_text(encoding="utf-8")

print(f"Nombre de caractères : {len(text)}")
```

```python
print(text[:3000])
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
)

chunks = text_splitter.split_text(text)
print(f"Nombre de chunks : {len(chunks)}")
```

```python
for i, chunk in enumerate(chunks[:5]):
    print(f"\n========== CHUNK {i} ==========\n")
    print(chunk)
```

```python
for i, chunk in enumerate(chunks[:10]):
    print(f"Chunk {i}: {len(chunk)} caractères")
```

```python
print("CHUNK 0:")
print(chunks[0])

print("\n\nCHUNK 1:")
print(chunks[1])
```

```python
from langchain_core.documents import Document

documents = []
for i, chunk in enumerate(chunks):
    document = Document(
        page_content=chunk,
        metadata={
            "source": markdown_file.name,
            "document_type": "course",
            "chunk_id": i,
        },
    )

    documents.append(document)
```

```python
print(documents[0].metadata)
```

```python
pdf_file = Path("../data/courses/introduction_ai.pdf")

documents = []

for i, chunk in enumerate(chunks):
    document = Document(
        page_content=chunk,
        metadata={
            "source": pdf_file.name,
            "document_type": "course",
            "chunk_id": i,
        },
    )

    documents.append(document)
```

```python
documents = []

for i, chunk in enumerate(chunks):
    document = Document(
        page_content=chunk,
        metadata={
            "source": pdf_file.name,
            "file_path": str(pdf_file),
            "document_type": "course",
            "chunk_id": i,
        },
    )

    documents.append(document)
```

```python
import json

chunks_data = [
    {
        "page_content": document.page_content,
        "metadata": document.metadata,
    }
    for document in documents
]
```

```python
chunks_file = Path("../data/processed/chunks.json")

chunks_file.write_text(
    json.dumps(
        chunks_data,
        indent=2,
        ensure_ascii=False
    ),
    encoding="utf-8"
)

print(f"Chunks sauvegardés dans : {chunks_file}")
```
