```python
def retrieve_documents(question: str, k: int = 5):
    """
    Retrieve the most relevant documents from the FAISS vector store.

    Args:
        question: User question.
        k: Number of documents to retrieve.

    Returns:
        List of relevant LangChain Documents.
    """
    
    results = loaded_vector_store.similarity_search(
        question,
        k=k
    )
    
    return results
```

```python
question = "Définir Internet"

results = retrieve_documents(question)

print(f"Nombre de documents récupérés : {len(results)}")
```

```python
for i, document in enumerate(results, start=1):
    print(f"\n{'=' * 60}")
    print(f"RESULTAT {i}")
    print(f"{'=' * 60}")
    
    print("\nContenu :")
    print(document.page_content)
    
    print("\nMetadata :")
    print(document.metadata)
```

```python
def retrieve_documents_with_scores(question: str, k: int = 5):
    """
    Retrieve relevant documents together with their similarity scores.
    """
    
    results = loaded_vector_store.similarity_search_with_score(
        question,
        k=k
    )
    
    return results
```

```python
question = "c'est quoi Internet"

results_with_scores = retrieve_documents_with_scores(
    question,
    k=5
)

for i, (document, score) in enumerate(results_with_scores, start=1):
    print(f"\n{'=' * 60}")
    print(f"RESULTAT {i}")
    print(f"{'=' * 60}")
    
    print(f"Score : {score}")
    print(f"Source : {document.metadata.get('source')}")
    print(f"Chunk ID : {document.metadata.get('chunk_id')}")
    
    print("\nContenu :")
    print(document.page_content[:500])
```

```python
def build_context(documents):
    """
    Combine retrieved documents into a single context string.
    """
    
    context_parts = []
    
    for i, document in enumerate(documents, start=1):
        source = document.metadata.get("source", "Unknown source")
        chunk_id = document.metadata.get("chunk_id", "Unknown")
        
        context_parts.append(
            f"[Document {i}]\n"
            f"Source: {source}\n"
            f"Chunk: {chunk_id}\n\n"
            f"{document.page_content}"
        )
    
    return "\n\n".join(context_parts)
```

```python
question = "C'est quoi Internet?"

documents = retrieve_documents(question, k=3)

context = build_context(documents)

print(context)
```

```python
def build_tutor_prompt(question: str, context: str) -> str:
    """
    Build the prompt used by the Tutor Agent.
    """
    
    return f"""
You are a learning tutor assisting a student.

Your task is to answer the student's question using the provided course context.

IMPORTANT RULES:
1. Use the course context as the primary source of information.
2. Do not invent information that is not supported by the context.
3. If the answer cannot be found in the context, clearly say that the information
   is not available in the provided course material.
4. Explain the answer clearly and pedagogically.
5. Do not mention information as coming from the course if it is not present in the context.

COURSE CONTEXT:
----------------
{context}
----------------

STUDENT QUESTION:
{question}

ANSWER:
"""
```

```python
def ask_tutor(question: str, k: int = 5):
    """
    Ask the Tutor a question using RAG.
    """
    
    # Step 1: Retrieve relevant documents
    documents = retrieve_documents(
        question,
        k=k
    )
    
    # Step 2: Build the context
    context = build_context(documents)
    
    # Step 3: Build the prompt
    prompt = build_tutor_prompt(
        question,
        context
    )
    
    # Step 4: Ask the LLM
    response = llm.invoke(prompt)
    
    return response.content, documents
```

```python
question = "C'est quoi internet?"

answer, documents = ask_tutor(
    question,
    k=5
)

print("=" * 70)
print("TUTOR ANSWER")
print("=" * 70)

print(answer)
```
