```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

print("RecursiveCharacterTextSplitter OK")
```

```python
from pathlib import Path

markdown_file = Path("../data/processed/chap1.md")

print(markdown_file.exists())
```

```python
text = markdown_file.read_text(encoding="utf-8")

print(f"Nombre de caractères : {len(text)}")
```

```python
print(text[:3000])
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
)

chunks = text_splitter.split_text(text)
print(f"Nombre de chunks : {len(chunks)}")
```

```python
for i, chunk in enumerate(chunks[:5]):
    print(f"\n========== CHUNK {i} ==========\n")
    print(chunk)
```

```python
for i, chunk in enumerate(chunks[:10]):
    print(f"Chunk {i}: {len(chunk)} caractères")
```

```python
print("CHUNK 0:")
print(chunks[0])

print("\n\nCHUNK 1:")
print(chunks[1])
```

```python
from langchain_core.documents import Document

documents = []
for i, chunk in enumerate(chunks):
    document = Document(
        page_content=chunk,
        metadata={
            "source": markdown_file.name,
            "document_type": "course",
            "chunk_id": i,
        },
    )

    documents.append(document)
```

```python
print(documents[0].metadata)
```

```python
pdf_file = Path("../data/courses/introduction_ai.pdf")

documents = []

for i, chunk in enumerate(chunks):
    document = Document(
        page_content=chunk,
        metadata={
            "source": pdf_file.name,
            "document_type": "course",
            "chunk_id": i,
        },
    )

    documents.append(document)
```

```python
documents = []

for i, chunk in enumerate(chunks):
    document = Document(
        page_content=chunk,
        metadata={
            "source": pdf_file.name,
            "file_path": str(pdf_file),
            "document_type": "course",
            "chunk_id": i,
        },
    )

    documents.append(document)
```

```python
import json

chunks_data = [
    {
        "page_content": document.page_content,
        "metadata": document.metadata,
    }
    for document in documents
]
```

```python
chunks_file = Path("../data/processed/chunks.json")

chunks_file.write_text(
    json.dumps(
        chunks_data,
        indent=2,
        ensure_ascii=False
    ),
    encoding="utf-8"
)

print(f"Chunks sauvegardés dans : {chunks_file}")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

print("HuggingFaceEmbeddings importé")
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
test_text = "Machine learning is a branch of artificial intelligence."

vector = embeddings.embed_query(test_text)

print(type(vector))
print(f"Dimension du vecteur : {len(vector)}")
```

```python
print(vector[:10])
```

```python
vector_1 = embeddings.embed_query(
    "What is machine learning?"
)

vector_2 = embeddings.embed_query(
    "Explain machine learning."
)

vector_3 = embeddings.embed_query(
    "How to cook pasta?"
)
```

```python
from langchain_community.vectorstores import FAISS

vector_store = FAISS.from_documents(
    documents,
    embeddings
)
```

```python
print(type(vector_store))
```

```python
from pathlib import Path

vectorstore_dir = Path("../vectorstore/courses")
vectorstore_dir.mkdir(
    parents=True,
    exist_ok=True
)

vector_store.save_local(
    str(vectorstore_dir)
)
```

```python
loaded_vector_store = FAISS.load_local(
    str(vectorstore_dir),
    embeddings,
    allow_dangerous_deserialization=True
)

print(type(loaded_vector_store))
```

```python
question = "définir Internet?"

results = loaded_vector_store.similarity_search(
    question,
    k=5
)

for i, result in enumerate(results):
    print(f"\n========== RESULT {i + 1} ==========\n")
    print(result.page_content)
    print("\nMetadata:")
    print(result.metadata)
```

```python
results_with_scores = loaded_vector_store.similarity_search_with_score(
    question,
    k=5
)

for document, score in results_with_scores:
    print("\nScore:", score)
    print(document.page_content[:500])
    print(document.metadata)
```
