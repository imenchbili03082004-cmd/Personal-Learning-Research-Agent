```python
from typing import Annotated

from langchain_core.tools import tool
from langchain_ollama import ChatOllama

print("Imports OK")
```

```python
@tool
def web_search(
    query: Annotated[str, "The search query"],
    max_results: Annotated[int, "Maximum number of results"] = 5,
):
    """
    Search the web and return a list of relevant results.
    """

    max_results = max(1, min(max_results, 5))

    response = ollama.web_search(
        query=query,
        max_results=max_results
    )

    if not response.results:
        return "No search results found."

    results = []

    for result in response.results:
        results.append(
            {
                "title": result.title,
                "url": result.url,
                "content": result.content,
            }
        )

    return results
```

```python
from ollama import Client

ollama = Client(
    host="https://ollama.com"
)

print("Ollama client OK")
```

```python
results = web_search.invoke(
    {
        "query": "artificial intelligence in education",
        "max_results": 3
    }
)

print(results)
```

```python
for i, result in enumerate(results, start=1):
    print("=" * 70)
    print(f"RESULT {i}")
    print("=" * 70)

    print("Title:", result["title"])
    print("URL:", result["url"])
    print("Content:")
    print(result["content"][:500])
    print()
```

```python
@tool
def web_fetch(
    url: Annotated[str, "The URL of the web page to fetch"]
):
    """
    Fetch the content of a web page.
    """

    response = ollama.web_fetch(
        url=url
    )

    return response
```

```python
url = results[0]["url"]

page = web_fetch.invoke(
    {
        "url": url
    }
)

print(page)
```

```python
llm = ChatOllama(
    base_url="https://api.ollama.com",
    model="gemma4:31b-cloud",
    temperature=0
)

print("LLM OK")
```

```python
tools = [
    web_search,
    web_fetch,
]

print(tools)
```

```python
list_of_tools = {
    tool.name: tool
    for tool in tools
}

print(list_of_tools.keys())
```

```python
llm_with_tools = llm.bind_tools(tools)

print("Tools bound to LLM")
```

```python
response = llm_with_tools.invoke(
    "Search the web for recent information about AI in education."
)

print(response)
```

```python
print(response.tool_calls)
```

```python
tool_call = response.tool_calls[0]

print("Tool:", tool_call["name"])
print("Arguments:", tool_call["args"])
```

```python
tool_result = list_of_tools[
    tool_call["name"]
].invoke(
    tool_call["args"]
)

print(tool_result)
```

```python
from langchain_core.messages import ToolMessage

tool_message = ToolMessage(
    content=str(tool_result),
    tool_call_id=tool_call["id"]
)

print(tool_message)
```

```python
from langchain.agents import create_agent
from langchain_ollama import ChatOllama

from tools import web_search, web_fetch

print("Imports OK")
```

```python
research_tools = [
    web_search,
    web_fetch,
]
```

```python
for tool in research_tools:
    print(tool.name)
```

```python
research_agent = create_agent(
    model=llm,
    tools=research_tools,
    system_prompt="""
You are a scientific research assistant.

Your role is to help students research scientific and
technical topics using reliable web sources.

You have access to two tools:

- web_search: search the web for relevant information.
- web_fetch: retrieve the content of a web page.

Rules:

1. Use web_search when the user asks for current,
   external, or scientific information that is not
   available in the student's course documents.

2. Use web_fetch when you need more information from
   a specific source found during the search.

3. Prefer relevant and trustworthy sources.

4. Clearly distinguish information obtained from web
   sources from information already available in the
   student's course material.

5. Do not invent sources, URLs, authors, or scientific
   results.

6. When possible, mention the sources used.

7. Give clear and structured answers suitable for a
   university student.
"""
)

print("Research Agent créé avec succès !")
```

```python
result = research_agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "What are the main applications of artificial intelligence in education?"
            }
        ]
    }
)
```

```python
def ask_research_agent(question: str):
    result = research_agent.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": question
                }
            ]
        }
    )

    return result
```

```python
result = ask_research_agent(
    "What are recent applications of AI in personalized education?"
)

print(result["messages"][-1].content)
```

```python
from langchain_core.messages import ToolMessage

def extract_tool_messages(result):
    tool_messages = []

    for message in result["messages"]:
        if isinstance(message, ToolMessage):
            tool_messages.append(message)

    return tool_messages

tool_messages = extract_tool_messages(result)

print(f"Nombre de ToolMessage : {len(tool_messages)}")
```

```python
for i, message in enumerate(tool_messages, start=1):

    print("=" * 70)
    print(f"TOOL RESULT {i}")
    print("=" * 70)

    print(message.content)
```
